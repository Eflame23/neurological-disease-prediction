```python
import torch
import torch.nn.functional as F
import joblib

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)
```

```python
MODEL_PATH = "/content/drive/MyDrive/AI_Neurological_Project/models/neurology_model_v1"
```

```python
label_encoder = joblib.load(
    "/content/drive/MyDrive/AI_Neurological_Project/label_encoder.pkl"
)

print("Loaded", len(label_encoder.classes_), "diseases")
```

```python
tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)

print("Tokenizer Loaded Successfully!")
```

```python
model = AutoModelForSequenceClassification.from_pretrained(MODEL_PATH)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model.to(device)

model.eval()

print("Using:", device)
```

```python
import torch
import torch.nn.functional as F

def predict_disease(symptoms, top_k=3):

    model.eval()

    inputs = tokenizer(
        symptoms,
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=128
    )

    device = model.device
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        outputs = model(**inputs)

    probabilities = F.softmax(outputs.logits, dim=1)

    top_probs, top_indices = torch.topk(probabilities, top_k)

    top_probs = top_probs.squeeze().cpu().numpy()
    top_indices = top_indices.squeeze().cpu().numpy()

    if top_k == 1:
        top_probs = [top_probs]
        top_indices = [top_indices]

    results = []

    medals = ["🥇", "🥈", "🥉"]

    print("=" * 75)
    print("🧠 AI Neurological Disease Prediction")
    print("=" * 75)

    print("\n📝 Patient Symptoms:\n")
    print(symptoms)

    print("\n" + "-" * 75)
    print("📊 Top Predictions\n")

    for i in range(top_k):

        disease = label_encoder.inverse_transform([int(top_indices[i])])[0]

        confidence = float(top_probs[i]) * 100

        results.append({
            "rank": i + 1,
            "disease": disease,
            "confidence": confidence
        })

        medal = medals[i] if i < 3 else f"{i+1}."

        print(f"{medal} {disease:<45} {confidence:6.2f}%")

    print("-" * 75)

    print("\n✅ Final Prediction")

    print(f"\nDisease   : {results[0]['disease']}")
    print(f"Confidence: {results[0]['confidence']:.2f}%")

    print("=" * 75)

    return results
```

```python
_ = predict_disease("""
I woke up today and couldn't smile properly. One side of my face feels weak and my eye won't close completely.
""")
```
